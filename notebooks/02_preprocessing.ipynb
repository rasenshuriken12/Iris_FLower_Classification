```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

from src.data_loader import load_raw_data
from src.preprocess import split_and_scale
```

## Why Split Before Scaling?

If we scale the full dataset, test data influences training → **data leakage**.
Always: `fit` on train, `transform` on test.

```python
df = load_raw_data(save_csv=False)
X_train, X_test, y_train, y_test, scaler, feature_names = split_and_scale(df)

print("Train:", X_train.shape, "Test:", X_test.shape)
print("Class distribution (train):")
print(y_train.value_counts(normalize=True))
```

✅ **Stratified split** preserves class proportions in both sets.

## Sanity Check: Scaled Data

```python
import numpy as np
print("Train mean:", np.round(X_train.mean(axis=0), 4))
print("Train std :", np.round(X_train.std(axis=0), 4))
```

Mean ≈ 0, Std ≈ 1 → scaling worked ✅
