# 01 — Exploratory Data Analysis (EDA)

**Goal:** Understand the Iris dataset before modeling.

**Questions to answer:**
1. What features exist? What's their range?
2. Are classes balanced?
3. Which features separate species best?
4. Any outliers or correlations?

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_raw_data

sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)
```

## 1. Load Data

We use `src/data_loader.py` for consistency across the project.

```python
df = load_raw_data(save_csv=True)
print("Shape:", df.shape)
df.head()
```

## 2. Basic Info

- **150 samples, 4 features, 1 target**
- Check dtypes, nulls, and summary statistics.

```python
df.info()
df.describe().T
```

**Observations:**
- No missing values ✅
- All features are `float64`
- Ranges: sepal length 4.3–7.9, petal length 1.0–6.9

## 3. Class Balance

```python
print(df["species"].value_counts())

df["species"].value_counts().plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Class Distribution")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

✅ **Balanced dataset:** 50 samples per species → no resampling needed.

## 4. Feature Distributions

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for i, col in enumerate(df.columns[:-1]):
    sns.histplot(data=df, x=col, hue="species", kde=True,
                 ax=axes[i//2, i%2], palette="husl")
plt.tight_layout()
plt.show()
```

**Observation:** Petal features show clear bimodal distributions → strong signal for classification.

## 5. Pairplot — Feature Relationships

```python
sns.pairplot(df, hue="species", palette="husl", diag_kind="kde")
plt.show()
```

**Key findings:**
- **Setosa** is linearly separable from the other two
- Versicolor & Virginica overlap slightly on sepal features
- **Petal length & width** are the most discriminative

## 6. Correlation Heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.drop(columns=["species"]).corr(),
            annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation")
plt.show()
```

**Observation:** Petal length & petal width = 0.96 correlation → high multicollinearity.

## 7. Outlier Check

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for i, col in enumerate(df.columns[:-1]):
    sns.boxplot(data=df, x="species", y=col, ax=axes[i//2, i%2], palette="husl")
plt.tight_layout()
plt.show()
```

## ✅ EDA Summary

| Finding | Implication |
|---------|-------------|
| Balanced classes | No resampling needed |
| No nulls | No imputation needed |
| Setosa linearly separable | Simple models should work |
| Petal features most useful | Feature selection may help |
| A few outliers in sepal width | Consider RobustScaler |
